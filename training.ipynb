```python
import lmdb
import pickle
from tqdm import tqdm

def save_complex_dataset_to_lmdb(dataset, lmdb_path):
    env = lmdb.open(lmdb_path, map_size=1e12)
    with env.begin(write=True) as txn:
        for idx in tqdm(range(len(dataset)), desc="Writing complex data to LMDB"):
            example = dataset[idx]  # può essere qualunque cosa serializzabile
            txn.put(f'{idx:08d}'.encode(), pickle.dumps(example))
        txn.put(b'__len__', pickle.dumps(len(dataset)))
```
